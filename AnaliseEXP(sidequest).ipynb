```python
import pandas as pd
import plotly.express as px
from ydata_profiling import ProfileReport
```

```python
df = pd.read_csv('padronizado.csv', decimal=',')
```

```python
aprovados = len([x for x in df['aprovado'] if x == 'Sim'])

reprovados = len([x for x in df['aprovado'] if x == 'Nao'])
```

```python
fig = px.pie(df, names='aprovado', title="distribuição de aprovados e reprovados", color='aprovado' , color_discrete_map={
        'Sim': 'green',
        'Nao': 'red',
       

    })
fig.update_traces(
    textinfo='value+percent'
)
fig.show()
```

```python
fig = px.histogram(df, x="sexo", title="Quantidade por Sexo")
fig.show()
```

```python
maioresNotasTabela = df.nlargest(5, 'Media').sort_values(by='Media')

fig = px.bar(maioresNotasTabela, x='Media', y='nome', title="Maiores Notas", orientation='h', color='Media', color_continuous_scale=px.colors.sequential.Plasma, text='Media')
fig.update_xaxes(range=[7.5, 8.6]) #para melhor ver as diferenças
fig.show()
```

```python
menoresNotasTabela = df.nsmallest(5, 'Media').sort_values(by='Media')

fig = px.bar(menoresNotasTabela, x='Media', y='nome', title="Maiores Notas", orientation='h', text='Media',
             color='Media', color_continuous_scale='Blues') 

fig.update_xaxes(range=[2.5, 3.6]) #tambem para melhor ver as diferenças
fig.show()
```

```python

maisAssiduos = df.nlargest(5, 'frequencia').sort_values(by='frequencia')


import plotly.express as px
fig = px.bar(maisAssiduos, 
             x='frequencia', 
             y='nome', 
             orientation='h', 
             title='Top 5 Alunos Mais Assíduos',
             text='frequencia',
             color='frequencia', 
             color_continuous_scale='Greens')



fig.update_xaxes(range=[95, 100])

fig.update_layout(yaxis=dict(title='Aluno'), xaxis=dict(title='Frequência (%)'))

fig.show()
```
